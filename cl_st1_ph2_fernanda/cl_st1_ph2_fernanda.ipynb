# Corpus Linguistics - Study 1 - Phase 2 - Fernanda

## Organise the documents and convert to raw text format

Rename the documents in corpus/00_sources/00_pdf_raw/ converting each filename to lowercase and replacing the spaces by an _

```python
import os

folder_path = 'corpus/00_sources/00_pdf_raw/'
for filename in os.listdir(folder_path):
    new_filename = filename.lower().replace(' ', '_')
    if new_filename != filename:
        os.rename(os.path.join(folder_path, filename), os.path.join(folder_path, new_filename))
```

Parse documents in corpus/00_sources/00_pdf_raw/ to corpus/00_sources/01_txt_raw/

```python
import os
import fitz  # PyMuPDF
import docx

input_dir = 'corpus/00_sources/00_pdf_raw/'
output_base_dir = 'corpus/00_sources/01_txt_raw/'
keywords = ['cpa', 'pdi', 'pga', 'raa', 'rade']

# Create output directories
for kw in keywords:
    os.makedirs(os.path.join(output_base_dir, kw), exist_ok=True)

for filename in os.listdir(input_dir):
    filepath = os.path.join(input_dir, filename)
    if not os.path.isfile(filepath):
        continue

    # Determine target directory based on keyword
    target_dir = None
    for kw in keywords:
        if kw in filename:
            target_dir = os.path.join(output_base_dir, kw)
            break

    if target_dir is None:
        continue

    base_name, ext = os.path.splitext(filename)
    output_filepath = os.path.join(target_dir, base_name + '.txt')

    try:
        if ext.lower() == '.pdf':
            doc = fitz.open(filepath)
            text = ""
            for page in doc:
                text += page.get_text()
            with open(output_filepath, 'w', encoding='utf-8') as f:
                f.write(text)
        elif ext.lower() == '.docx':
            doc = docx.Document(filepath)
            text = '\n'.join([para.text for para in doc.paragraphs])
            with open(output_filepath, 'w', encoding='utf-8') as f:
                f.write(text)
    except Exception as e:
        print(f"Error processing {filename}: {e}")
```
